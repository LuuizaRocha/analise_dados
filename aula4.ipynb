```python
# Carregar dados e agregar por estado
import pandas as pd

df = pd.read_excel("CD2022_Populacao_2010_Compatibilizada_20231222.xlsx", sheet_name="Municípios", header=2)
df = df.drop(columns=[df.columns[0]])
df = df.dropna(subset=["UF", "NOME DO MUNICÍPIO"])
df = df.rename(columns={
    "População 2010 (Alterações de Limites até 2022)1": "Pop_2010_Compat",
    "População Censo 2022": "Pop_2022"
})
df["Pop_2010_Compat"] = pd.to_numeric(df["Pop_2010_Compat"], errors="coerce")
df["Pop_2022"] = pd.to_numeric(df["Pop_2022"], errors="coerce")

tabela_estado = df.groupby("UF", as_index=False).agg(
    Pop_2010=("Pop_2010_Compat", "sum"),
    Pop_2022=("Pop_2022", "sum")
)
```

```python
# Ordenar por crescimento
tabela_estado["Diferenca"] = tabela_estado["Pop_2022"] - tabela_estado["Pop_2010"]
tabela_estado["Crescimento_%"] = (tabela_estado["Diferenca"] / tabela_estado["Pop_2010"]) * 100
tabela_estado = tabela_estado.sort_values("Diferenca", ascending=False).reset_index(drop=True)
```

```python
# Salvar em csv
tabela_estado.to_csv("populacao_por_estado_2010_2022.csv", index=False, encoding="utf-8-sig")
```

```python
# Trabalhar por municipio
tabela_municipio = df[["UF", "COD. UF", "COD. MUNIC", "NOME DO MUNICÍPIO", "Pop_2010_Compat", "Pop_2022"]].copy()
tabela_municipio = tabela_municipio.rename(columns={"Pop_2010_Compat": "Pop_2010", "NOME DO MUNICÍPIO": "Municipio"})
tabela_municipio["Diferenca"] = tabela_municipio["Pop_2022"] - tabela_municipio["Pop_2010"]
tabela_municipio["Crescimento_%"] = (tabela_municipio["Diferenca"] / tabela_municipio["Pop_2010"]) * 100
```

```python
# Ordenar por crescimento e salvar
tabela_municipio = tabela_municipio.sort_values("Diferenca", ascending=False).reset_index(drop=True)
tabela_municipio.to_csv("populacao_por_municipio_2010_2022.csv", index=False, encoding="utf-8-sig")
```
